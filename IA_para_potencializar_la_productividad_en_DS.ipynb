El código:

```python
import plotly.graph_objects as go

# Asegurarse de que el dataframe de ranking_paises_reciente está disponible
if 'ranking_paises_reciente' not in locals():
    # Definir las columnas de los años de interés
    años_recientes = [str(año) for año in range(2010, 2025)]
    # Sumar la producción de energía para los años recientes por país
    produccion_reciente_por_pais = df_produccion[['pais'] + años_recientes].copy()
    produccion_reciente_por_pais['produccion_total_reciente'] = produccion_reciente_por_pais[años_recientes].sum(axis=1)
    # Ranquear los países por producción total en los años recientes
    ranking_paises_reciente = produccion_reciente_por_pais.sort_values('produccion_total_reciente', ascending=False).head(10).reset_index(drop=True)


# Seleccionar solo las columnas de país y años para la visualización
produccion_anos = ranking_paises_reciente[['pais'] + años_recientes].copy()

# Transformar los datos de formato "wide" a formato "long"
produccion_long = produccion_anos.melt(
    id_vars='pais',
    var_name='Año',
    value_name='Producción de Energía (cuatrillones de BTU)'
)

# Convertir la columna 'Año' a tipo numérico
produccion_long['Año'] = produccion_long['Año'].astype(int)

# Crear el gráfico de líneas con Plotly
fig = go.Figure()

for pais in produccion_long['pais'].unique():
    df_pais = produccion_long[produccion_long['pais'] == pais]
    fig.add_trace(go.Scatter(
        x=df_pais['Año'],
        y=df_pais['Producción de Energía (cuatrillones de BTU)'],
        mode='lines',
        name=pais
    ))

# Configurar el layout del gráfico
fig.update_layout(
    title='Historial de Producción de Energía por País (2010-2024)',
    xaxis_title='Año',
    yaxis_title='Producción de Energía (cuatrillones de BTU)',
    hovermode='x unified' # Muestra información para todos los países en una fecha dada al pasar el ratón
)

# Mostrar el gráfico
fig.show()
```
